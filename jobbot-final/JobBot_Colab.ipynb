```python
#@title 1) Upload jobbot-repo-final.zip, unpack and install (3-5 min first time)
import os, sys, zipfile, glob
from google.colab import files
if not glob.glob("/content/jobbot-final/app.py"):
    up = files.upload()
    zipfile.ZipFile([k for k in up if k.endswith(".zip")][0]).extractall("/content")
%cd /content/jobbot-final
sys.path.insert(0, "/content/jobbot-final")
!pip install -q pypdf reportlab pyyaml requests playwright
!playwright install --with-deps chromium
print("READY")
```

```python
#@title 2) Your settings
mode = "find" #@param ["find", "dry", "live"]
max_applications = 15 #@param {type:"integer"}
min_match_score = 40 #@param {type:"integer"}
max_years_required = 1 #@param [0, 1, 2, 3] {type:"raw"}
per_company = 2 #@param [1, 2, 3] {type:"raw"}
pause_between_seconds = 10 #@param {type:"integer"}

linkedin = "" #@param {type:"string"}
github = "" #@param {type:"string"}
city = "India" #@param {type:"string"}
current_org = "" #@param {type:"string"}
notice_period = "Immediate" #@param {type:"string"}
expected_ctc = "" #@param {type:"string"}
years_experience = "1" #@param {type:"string"}
graduation_year = "" #@param {type:"string"}
degree = "" #@param {type:"string"}
university = "" #@param {type:"string"}
cgpa = "" #@param {type:"string"}
preferred_location = "" #@param {type:"string"}

extra_company_pages = "" #@param {type:"string"}
groq_api_key = "" #@param {type:"string"}
gmail_address = "" #@param {type:"string"}
gmail_app_password = "" #@param {type:"string"}
save_history_to_google_drive = False #@param {type:"boolean"}
```

```python
#@title 3) Upload your resume (PDF)
from google.colab import files
r = files.upload()
RESUME = "/content/" + [k for k in r if k.lower().endswith(".pdf")][0]
print("Using", RESUME)
```

```python
#@title 4) RUN (watch the live log)
import jb_colab
out = "/content/jobbot_out"
if save_history_to_google_drive:
    from google.colab import drive
    drive.mount("/content/drive"); out = "/content/drive/MyDrive/jobbot_out"
candidate = {"linkedin": linkedin, "github": github, "city": city, "current_company": current_org,
             "notice_period": notice_period, "expected_ctc": expected_ctc, "years_experience": years_experience,
             "graduation_year": graduation_year, "degree": degree, "university": university, "cgpa": cgpa,
             "preferred_location": preferred_location}
db, out_dir = jb_colab.run(RESUME, mode=mode, candidate=candidate, max_apps=max_applications,
        min_score=min_match_score, max_years=int(max_years_required), per_company=int(per_company),
        extra_boards=extra_company_pages.replace(",", "\n"), mail_user=gmail_address, mail_pass=gmail_app_password,
        pause=(pause_between_seconds, pause_between_seconds + 10), groq_key=groq_api_key, out_dir=out)
```

```python
#@title 5) Dashboard
jb_colab.show(out_dir)
```

```python
#@title 6) Download dashboard, CSV and tailored resumes
jb_colab.download(out_dir)
```
